````python
import json
import polyglot_stream_heal as psh

# 1. Incremental Streaming JSON + Mermaid Healing
t = psh.IncrementalTransducer(max_table_cols=4, max_chart_bars=8)
for chunk in [
    '### Live Stream\n```json\n',
    '{"service": "gateway", "active": tru, "p99_ms": 6.',
]:
    t.push_chunk(chunk)

print('Active Healed JSON:', json.loads(t.snapshot_active_json())['content'])

# 2. Context-Free Mermaid Dyck Node & Subgraph Healing
raw_mermaid = 'flowchart LR\n  subgraph Compute (v2)\n    A[Evaluate O(f(g(h(x))))] -- stream --> B[(DB (v16))]'
print('\nHealed Mermaid:\n' + psh.heal_mermaid(raw_mermaid))
````
